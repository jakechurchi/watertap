# Figures for the CrossCutting Paper

```python
import matplotlib.pyplot as plt

# Breakdown of the weekly OPEX savings by type
# Data = [baseline, flexible]
fixed_demand = [4931, 5590]
peak_demand = [9095, 5989]
energy = [20627, 19870]
feed = [50843, 50997]
brine = [15304, 15718]
chemicals = [19298, 19330]

categories = {
    "Fixed demand": fixed_demand,
    "Peak demand": peak_demand,
    "Energy": energy,
    "Feed": feed,
    "Brine": brine,
    "Chemicals": chemicals,
}

labels = ["Baseline", "Flexible"]
x = range(len(labels))

fig, ax = plt.subplots(figsize=(7, 7))

bottom = [0, 0]
for name, values in categories.items():
    ax.bar(x, values, bottom=bottom, label=name)
    bottom = [bottom[i] + values[i] for i in range(len(values))]

ax.set_xticks(list(x))
ax.set_xticklabels(labels)
ax.set_ylabel("Weekly Cost (USD)")
ax.set_title("Weekly OPEX Breakdown")
ax.legend(title="Cost Type", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# Weekly Electricity OPEX (first three cost categories)
electricity_categories = {
    "Fixed demand": fixed_demand,
    "Peak demand": peak_demand,
    "Energy": energy,
}

fig, ax = plt.subplots(figsize=(7, 7))

bottom = [0, 0]
for name, values in electricity_categories.items():
    ax.bar(x, values, bottom=bottom, label=name)
    bottom = [bottom[i] + values[i] for i in range(len(values))]

ax.set_xticks(list(x))
ax.set_xticklabels(labels)
ax.set_ylabel("Weekly Cost (USD)")
ax.set_title("Weekly Electricity OPEX")
ax.legend(title="Cost Type", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# Weekly Flow OPEX (last three cost categories)
electricity_categories = {
    "Feed": feed,
    "Brine": brine,
    "Chemicals": chemicals,
}

fig, ax = plt.subplots(figsize=(7, 7))

bottom = [0, 0]
for name, values in electricity_categories.items():
    ax.bar(x, values, bottom=bottom, label=name)
    bottom = [bottom[i] + values[i] for i in range(len(values))]

ax.set_xticks(list(x))
ax.set_xticklabels(labels)
ax.set_ylabel("Weekly Cost (USD)")
ax.set_title("Weekly Flow OPEX")
ax.legend(title="Cost Type", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# Grouped bar chart: baseline vs flexible for each cost category
category_names = list(categories.keys())
baseline_values = [categories[name][0] for name in category_names]
flexible_values = [categories[name][1] for name in category_names]

x_pos = list(range(len(category_names)))
bar_width = 0.4

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar([i - bar_width / 2 for i in x_pos], baseline_values, width=bar_width, label="baseline")
ax.bar([i + bar_width / 2 for i in x_pos], flexible_values, width=bar_width, label="flexible")

ax.set_xticks(x_pos)
ax.set_xticklabels(category_names, rotation=30, ha="right")
ax.set_xlabel("Cost Category")
ax.set_ylabel("Weekly Cost (USD)")
ax.set_title("Weekly OPEX by Cost Type")
ax.legend()

plt.tight_layout()
plt.show()
```
